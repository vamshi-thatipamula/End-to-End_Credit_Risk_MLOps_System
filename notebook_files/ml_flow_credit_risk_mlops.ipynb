### Experiment 8: Logistic Regression with SMOTE-Tomek and Optuna

```python
import optuna
from sklearn.metrics import make_scorer, f1_score
from sklearn.model_selection import cross_val_score

# --------------------------------------------------
# Define Optuna objective function
# --------------------------------------------------

def objective(trial):

    # Define hyperparameter search space

    params = {
        "C": trial.suggest_float(
            "C",
            1e-4,
            1e4,
            log=True
        ),

        "solver": trial.suggest_categorical(
            "solver",
            [
                "lbfgs",
                "liblinear",
                "saga",
                "newton-cg"
            ]
        ),

        "tol": trial.suggest_float(
            "tol",
            1e-6,
            1e-1,
            log=True
        ),

        "class_weight": trial.suggest_categorical(
            "class_weight",
            [
                None,
                "balanced"
            ]
        )
    }


    # Initialize Logistic Regression

    model = LogisticRegression(
        **params,
        max_iter=10000,
        random_state=42
    )


    # Cross-validation using SMOTE-Tomek data

    scores = cross_val_score(
        model,
        X_train_smt,
        y_train_smt,
        cv=3,
        scoring="f1",
        n_jobs=-1
    )


    return scores.mean()


# --------------------------------------------------
# Create Optuna study
# --------------------------------------------------

study_logistic = optuna.create_study(
    direction="maximize"
)


# --------------------------------------------------
# Perform Optuna optimization
# --------------------------------------------------

study_logistic.optimize(
    objective,
    n_trials=50
)


# --------------------------------------------------
# Display best trial
# --------------------------------------------------

print("Best Trial:")

best_trial = study_logistic.best_trial

print(
    f"F1 Score: {best_trial.value:.4f}"
)

print("Best Hyperparameters:")

for key, value in best_trial.params.items():

    print(
        f"  {key}: {value}"
    )


# --------------------------------------------------
# Create best Logistic Regression model
# --------------------------------------------------

best_model_logistic = LogisticRegression(
    **study_logistic.best_params,
    max_iter=10000,
    random_state=42
)


# --------------------------------------------------
# Train best model
# --------------------------------------------------

best_model_logistic.fit(
    X_train_smt,
    y_train_smt
)


# --------------------------------------------------
# Generate predictions
# --------------------------------------------------

y_pred = best_model_logistic.predict(
    X_test_encoded
)


# --------------------------------------------------
# Display classification report
# --------------------------------------------------

print("Classification Report:")

print(
    classification_report(
        y_test,
        y_pred
    )
)


# Generate report dictionary

report = classification_report(
    y_test,
    y_pred,
    output_dict=True
)


# --------------------------------------------------
# MLflow Experiment Tracking
# --------------------------------------------------

with mlflow.start_run(
    run_name=(
        "Experiment 8 - "
        "Logistic Regression with "
        "SMOTE-Tomek and Optuna"
    )
):

    # ----------------------------------------------
    # Log model information
    # ----------------------------------------------

    mlflow.log_param(
        "model",
        "Logistic Regression"
    )

    mlflow.log_param(
        "sampling_method",
        "SMOTE-Tomek"
    )

    mlflow.log_param(
        "optimization_method",
        "Optuna"
    )


    # ----------------------------------------------
    # Log Optuna hyperparameters
    # ----------------------------------------------

    mlflow.log_params(
        study_logistic.best_params
    )


    # ----------------------------------------------
    # Log Optuna best CV score
    # ----------------------------------------------

    mlflow.log_metric(
        "best_cv_f1_score",
        study_logistic.best_value
    )


    # ----------------------------------------------
    # Log tags
    # ----------------------------------------------

    mlflow.set_tag(
        "model_type",
        "Logistic Regression"
    )

    mlflow.set_tag(
        "sampling_method",
        "SMOTE-Tomek"
    )

    mlflow.set_tag(
        "optimization_method",
        "Optuna"
    )


    # ----------------------------------------------
    # Log test-set metrics
    # ----------------------------------------------

    mlflow.log_metric(
        "accuracy",
        report["accuracy"]
    )

    # Non-default class (0)

    mlflow.log_metric(
        "precision_non_default",
        report["0"]["precision"]
    )

    mlflow.log_metric(
        "recall_non_default",
        report["0"]["recall"]
    )

    mlflow.log_metric(
        "f1_non_default",
        report["0"]["f1-score"]
    )

    # Default class (1)

    mlflow.log_metric(
        "precision_default",
        report["1"]["precision"]
    )

    mlflow.log_metric(
        "recall_default",
        report["1"]["recall"]
    )

    mlflow.log_metric(
        "f1_default",
        report["1"]["f1-score"]
    )


    # ----------------------------------------------
    # Log model artifact
    # ----------------------------------------------

    mlflow.sklearn.log_model(
        sk_model=best_model_logistic,
        name="model"
    )
```

### Experiment 9: Random Forest with SMOTE-Tomek and Optuna

```python
# --------------------------------------------------
# Define Optuna objective function
# --------------------------------------------------

def objective(trial):

    # Define hyperparameter search space

    params = {
        "n_estimators": trial.suggest_int(
            "n_estimators",
            100,
            300,
            step=100
        ),

        "max_depth": trial.suggest_int(
            "max_depth",
            5,
            20
        ),

        "min_samples_split": trial.suggest_int(
            "min_samples_split",
            2,
            5
        ),

        "min_samples_leaf": trial.suggest_int(
            "min_samples_leaf",
            1,
            3
        ),

        "max_features": trial.suggest_categorical(
            "max_features",
            ["sqrt"]
        ),

        "bootstrap": trial.suggest_categorical(
            "bootstrap",
            [True]
        ),

        "class_weight": trial.suggest_categorical(
            "class_weight",
            [None, "balanced"]
        )
    }


    # Initialize Random Forest

    model = RandomForestClassifier(
        **params,
        random_state=42,
        n_jobs=-1
    )


    # Cross-validation using SMOTE-Tomek data

    scores = cross_val_score(
        model,
        X_train_smt,
        y_train_smt,
        cv=2,
        scoring="f1",
        n_jobs=-1
    )


    return scores.mean()


# --------------------------------------------------
# Create Optuna study
# --------------------------------------------------

study_rf = optuna.create_study(
    direction="maximize",
    sampler=optuna.samplers.TPESampler(
        seed=42
    )
)


# --------------------------------------------------
# Perform Optuna optimization
# --------------------------------------------------

study_rf.optimize(
    objective,
    n_trials=20
)


# --------------------------------------------------
# Display best trial
# --------------------------------------------------

print("Best Trial:")

best_trial = study_rf.best_trial

print(
    f"F1 Score: {best_trial.value:.4f}"
)

print("Best Hyperparameters:")

for key, value in best_trial.params.items():

    print(
        f"  {key}: {value}"
    )


# --------------------------------------------------
# Create best Random Forest model
# --------------------------------------------------

best_model_rf = RandomForestClassifier(
    **study_rf.best_params,
    random_state=42,
    n_jobs=-1
)


# --------------------------------------------------
# Train best model using SMOTE-Tomek data
# --------------------------------------------------

best_model_rf.fit(
    X_train_smt,
    y_train_smt
)


# --------------------------------------------------
# Generate predictions
# --------------------------------------------------

y_pred = best_model_rf.predict(
    X_test_encoded
)


# --------------------------------------------------
# Display classification report
# --------------------------------------------------

print("Classification Report:")

print(
    classification_report(
        y_test,
        y_pred
    )
)


# Generate report dictionary

report = classification_report(
    y_test,
    y_pred,
    output_dict=True
)


# --------------------------------------------------
# MLflow Experiment Tracking
# --------------------------------------------------

with mlflow.start_run(
    run_name=(
        "Experiment 9 - "
        "Random Forest with "
        "SMOTE-Tomek and Optuna"
    )
):

    # ----------------------------------------------
    # Log model information
    # ----------------------------------------------

    mlflow.log_param(
        "model",
        "Random Forest"
    )

    mlflow.log_param(
        "sampling_method",
        "SMOTE-Tomek"
    )

    mlflow.log_param(
        "optimization_method",
        "Optuna"
    )


    # ----------------------------------------------
    # Log Optuna hyperparameters
    # ----------------------------------------------

    mlflow.log_params(
        study_rf.best_params
    )


    # ----------------------------------------------
    # Log Optuna best CV score
    # ----------------------------------------------

    mlflow.log_metric(
        "best_cv_f1_score",
        study_rf.best_value
    )


    # ----------------------------------------------
    # Log tags
    # ----------------------------------------------

    mlflow.set_tag(
        "model_type",
        "Random Forest"
    )

    mlflow.set_tag(
        "sampling_method",
        "SMOTE-Tomek"
    )

    mlflow.set_tag(
        "optimization_method",
        "Optuna"
    )


    # ----------------------------------------------
    # Log test-set metrics
    # ----------------------------------------------

    mlflow.log_metric(
        "accuracy",
        report["accuracy"]
    )

    # Non-default class (0)

    mlflow.log_metric(
        "precision_non_default",
        report["0"]["precision"]
    )

    mlflow.log_metric(
        "recall_non_default",
        report["0"]["recall"]
    )

    mlflow.log_metric(
        "f1_non_default",
        report["0"]["f1-score"]
    )

    # Default class (1)

    mlflow.log_metric(
        "precision_default",
        report["1"]["precision"]
    )

    mlflow.log_metric(
        "recall_default",
        report["1"]["recall"]
    )

    mlflow.log_metric(
        "f1_default",
        report["1"]["f1-score"]
    )


    # ----------------------------------------------
    # Log model artifact
    # ----------------------------------------------

    mlflow.sklearn.log_model(
        sk_model=best_model_rf,
        name="model"
    )
```

```python
# Step 1: Find Experiment 8 and its Logged Model

import mlflow
from mlflow.tracking import MlflowClient

# Create MLflow client
client = MlflowClient()

# Get the Credit Risk experiment
experiment = client.get_experiment_by_name(
    "Credit Risk Modelling Experiments"
)

print("Experiment ID:", experiment.experiment_id)

# Find Experiment 8
runs = client.search_runs(
    experiment_ids=[experiment.experiment_id],
    filter_string=(
        "attributes.run_name = "
        "'Experiment 8 - Logistic Regression with SMOTE-Tomek and Optuna'"
    ),
    order_by=[
        "attributes.start_time DESC"
    ]
)

if not runs:
    raise ValueError(
        "Experiment 8 run was not found."
    )

run = runs[0]

run_id = run.info.run_id

print("Experiment 8 Run ID:", run_id)
```

```python
# Step 2: Find the Logged Model

# Find the Logged Model created by Experiment 8

logged_models = mlflow.search_logged_models(
    filter_string=f"source_run_id = '{run_id}'"
)

print(
    "Number of Logged Models:",
    len(logged_models)
)

display(logged_models)
```

```python
# Step 3: Extract the Model ID

# Get the first Logged Model

logged_model = logged_models.iloc[0]

model_id = logged_model["model_id"]

print("Logged Model ID:", model_id)
print("Model Name:", logged_model["name"])
print("Source Run ID:", logged_model["source_run_id"])
print("Artifact Location:", logged_model["artifact_location"])
```

```python
# Step 4: Register the DagsHub Logged Model

# Register Experiment 8 Model

registered_model_name = "Credit Risk Model"

model_uri = f"models:/{model_id}"

registered_model = mlflow.register_model(
    model_uri=model_uri,
    name=registered_model_name
)

print("Model registered successfully.")
print("Model name:", registered_model.name)
print("Model version:", registered_model.version)
```

```python
# Step 5: Load the registered model

model_name = "Credit Risk Model"
model_version = 1

model_uri = f"models:/{model_name}/{model_version}"

loaded_model = mlflow.sklearn.load_model(
    model_uri
)

print("Registered Model URI:")
print(model_uri)

print("\nModel loaded successfully.")
print("Model type:", type(loaded_model))
```
